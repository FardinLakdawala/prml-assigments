# Question 1: Central Limit Theorem Demonstration with Die Rolls
This experiment rolls a fair six-sided die multiple times (from $N=1$ to $N=50$ rolls), sums the outcomes, and repeats this process for different experiment counts (trials). We analyze how the distribution of the sum approaches a normal distribution as $N$ increases, and calculate the empirical mean and variance.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Purpose: Demonstrate the Central Limit Theorem by simulating die rolls.
# Working:
# 1. We simulate rolling a die N times (N ranges from 1 to 50).
# 2. We repeat this experiment for different trial sizes: 100, 500, 1000, 5000, 10000, 50000, 100000.
# 3. For a given N and trial size, we compute the sum of the rolls, calculate mean/variance, and plot histograms.
# 4. Due to the high number of combinations (50 choices of N * 7 trial sizes = 350 plots), we will select representative values of N
#    (e.g., N = 1, 2, 5, 10, 30, 50) and plot/save their distributions to show the transition to normality.

trials_list = [100, 500, 1000, 5000, 10000, 50000, 100000]
selected_N = [1, 2, 5, 10, 30, 50] # Representative number of dice rolls
os.makedirs('die_plots', exist_ok=True)

for N in selected_N:
    fig, axes = plt.subplots(4, 2, figsize=(14, 18))
    axes = axes.flatten()
    fig.suptitle(f'Sum of Rolling {N} Dice across Different Trial Sizes', fontsize=16)

    for idx, trials in enumerate(trials_list):
        # Roll a fair 6-sided die: values 1 to 6. Shape: (trials, N)
        rolls = np.random.randint(1, 7, size=(trials, N))
        # Calculate the sum of rolls for each trial
        sums = np.sum(rolls, axis=1)

        # Compute empirical mean and variance
        mean_val = np.mean(sums)
        var_val = np.var(sums)

        # Plot the distribution
        ax = axes[idx]
        sns.histplot(sums, bins=range(N, 6*N + 2), kde=True, ax=ax, color='skyblue', stat='density')
        ax.set_title(f'{trials} Trials (Mean: {mean_val:.2f}, Var: {var_val:.2f})')
        ax.set_xlabel('Sum of Rolls')
        ax.set_ylabel('Density')

        # Print statistics to console
        print(f"N={N} rolls, Trials={trials:7d} -> Empirical Mean: {mean_val:8.4f} (Theoretical: {N*3.5:8.4f}), Empirical Var: {var_val:8.4f} (Theoretical: {N*2.9167:8.4f})")

    # Hide the 8th unused subplot
    axes[-1].axis('off')
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.savefig(f'die_plots/die_roll_N_{N}.png')
    plt.show()
```

# Question 2: Binomial Distribution & Estimating E-Commerce Success Rate
Based on observing 97 positive reviews out of 100, we want to construct binomial distributions under different hypothetical success rates ($p$) to see which success rates make our observation of 97/100 highly probable.

```python
# Purpose: Estimate the success rate parameter (p) of a Binomial distribution given an observed sample (97 successes in 100 trials).
# Working:
# 1. We test different candidate success rates (p) ranging from 0.85 to 0.99.
# 2. For each candidate p, we simulate a Binomial process (n=100) repeated 10,000 times.
# 3. We plot the resulting distributions and highlight our observed value of 97 to visually identify the most likely success rate.

n = 100
observed_successes = 97
candidate_rates = [0.90, 0.94, 0.97, 0.99]
simulations = 50000

plt.figure(figsize=(15, 10))
for i, p in enumerate(candidate_rates, 1):
    # Draw samples from the Binomial distribution
    samples = np.random.binomial(n, p, simulations)

    plt.subplot(2, 2, i)
    # Compute the proportion of times we get exactly 97 successes
    probability_of_97 = np.mean(samples == observed_successes)

    sns.histplot(samples, bins=np.arange(min(samples)-0.5, max(samples)+1.5, 1), stat='probability', color='teal', alpha=0.7)
    plt.axvline(observed_successes, color='red', linestyle='--', linewidth=2, label=f'Observed (97)')
    plt.title(f'Success Rate p = {p} (Prob of getting 97: {probability_of_97:.4f})')
    plt.xlabel('Number of Positive Reviews')
    plt.ylabel('Probability')
    plt.legend()

plt.suptitle('Evaluating E-Commerce Product Success Rate Candidates', fontsize=16)
plt.tight_layout()
plt.show()
```

# Question 3: Entropy Calculation for Continuous Variables
For the dataset `/content/advertising.csv`, we calculate Shannon entropy. Since the values are continuous, we first bin them into discrete intervals before calculating entropy.

```python
import pandas as pd

# Purpose: Calculate the Shannon Entropy for each column in advertising.csv.
# Working:
# 1. Load the dataset using pandas.
# 2. Define a function to compute Shannon entropy: H(X) = -sum(P(x) * log2(P(x))).
# 3. If a column contains continuous values, we bin it into 10 equal-width intervals using pd.cut to make it discrete.
# 4. Calculate and display the entropy for each column.

def calculate_entropy(series, is_continuous=True, num_bins=10):
    # Bin the data if it is continuous
    if is_continuous:
        discretized = pd.cut(series, bins=num_bins)
    else:
        discretized = series

    # Calculate probability distribution
    prob_dist = discretized.value_counts(normalize=True)
    # Compute Shannon entropy
    entropy = -np.sum(prob_dist * np.log2(prob_dist + 1e-12)) # 1e-12 added to avoid log(0)
    return entropy

# Load advertising dataset
adv_df = pd.read_csv('/content/advertising.csv')
print("Dataset Sample:")
display(adv_df.head())

print("\n--- Entropy for Each Column ---")
for col in adv_df.columns:
    # Assuming all numerical features in this advertising dataset are continuous
    is_cont = pd.api.types.is_numeric_dtype(adv_df[col])
    entropy_val = calculate_entropy(adv_df[col], is_continuous=is_cont)
    print(f"Column: {col:15s} | Continuous: {str(is_cont):5s} | Shannon Entropy: {entropy_val:.4f} bits")
```

# Question 4: Find-S Algorithm (Maximal Specific Hypothesis)
Using the dataset `/content/enjoy_sport.csv`, we find the maximally specific hypothesis that fits all positive instances using the Find-S concept.

```python
# Purpose: Find the maximally specific hypothesis for the enjoy_sport.csv dataset using the Find-S Algorithm.
# Working:
# 1. Load enjoy_sport.csv dataset.
# 2. Initialize the hypothesis with the most specific possible values: ['?', '?', '?', '?', '?', '?'] is updated to the first positive instance.
# 3. For each subsequent positive training example, generalize the hypothesis attribute if it doesn't match the instance value (replace with '?').
# 4. Negative instances are ignored by the Find-S algorithm.

# Load the dataset
sport_df = pd.read_csv('/content/enjoy_sport.csv')
print("Enjoy Sport Dataset:")
display(sport_df)

# Separate features and target label (assuming the last column is the class label)
features = np.array(sport_df.iloc[:, :-1])
target = np.array(sport_df.iloc[:, -1])

# Initialize specific hypothesis with the first positive training instance
specific_h = None
for idx, val in enumerate(target):
    if val.lower() in ['yes', '1', 'true']:
        specific_h = features[idx].copy()
        break

print(f"\nInitial Specific Hypothesis: {specific_h}")

# Refine hypothesis using remaining positive instances
for idx, val in enumerate(target):
    if val.lower() in ['yes', '1', 'true']:
        for x in range(len(specific_h)):
            if features[idx][x] != specific_h[x]:
                specific_h[x] = '?'

print(f"Final Maximally Specific Hypothesis (Find-S): {specific_h}")
```

# Question 5: Candidate Elimination Algorithm (Version Space)
Using the dataset `/content/enjoy_sport.csv`, we compute the General ($G$) and Specific ($S$) boundaries of the version space.

```python
# Purpose: Find the version space (Specific boundary 'S' and General boundary 'G') using the Candidate Elimination algorithm.
# Working:
# 1. Initialize S to the first positive example and G to contain all '?' elements.
# 2. Iterate through each training example.
# 3. For positive examples: Generalize S to fit the example, and remove any hypotheses from G that do not cover the example.
# 4. For negative examples: Specialize G to exclude the example while ensuring elements in G remain consistent with S.

num_attributes = features.shape[1]

# Initialize S (most specific) and G (most general)
S = ['0'] * num_attributes
G = [['?'] * num_attributes]

# Find first positive sample to properly initialize S
for idx, val in enumerate(target):
    if val.lower() in ['yes', '1', 'true']:
        S = list(features[idx])
        break

print("Starting Candidate Elimination...")
print(f"Initial S: {S}")
print(f"Initial G: {G}\n")

for idx, val in enumerate(target):
    is_positive = val.lower() in ['yes', '1', 'true']
    current_instance = features[idx]

    if is_positive:
        # For positive instance: generalize S, filter G
        for i in range(num_attributes):
            if current_instance[i] != S[i]:
                S[i] = '?'

        # Remove hypotheses from G that are not consistent with the positive instance
        G = [g for g in G if all(g[j] == '?' or g[j] == current_instance[j] for j in range(num_attributes))]

    else:
        # For negative instance: specialize G
        temp_G = []
        for g in G:
            # If g is consistent with the negative example, we must specialize it
            if any(g[j] != '?' and g[j] != current_instance[j] for j in range(num_attributes)):
                temp_G.append(g)
            else:
                # Generate specialized versions of g
                for j in range(num_attributes):
                    if g[j] == '?' and S[j] != '?' and S[j] != current_instance[j]:
                        special_g = g.copy()
                        special_g[j] = S[j]
                        if special_g not in temp_G:
                            temp_G.append(special_g)
        G = temp_G

# Print final boundaries defining the Version Space
print("Candidate Elimination Process Complete.")
print(f"Final Specific Boundary (S): {S}")
print(f"Final General Boundary (G): {G}")
```
